## Our Project Work

### The Goal

In this project, we take on the role of Business Analysts and database developers\. Our task is to design and implement a relational database system for a hotel reservation system used in Swiss hotels and ski resorts\.

The focus of this project is entirely on the management, organisation and processing of data\. Functional aspects such as website design, login systems or user experience are intentionally excluded from the scope of this module\.

The main goal of this project is to create a fully functional relational database for a hotel reservation system\.

The system must be capable of storing and managing:

- hotels and owners

- rooms and room types

- customers and addresses

- bookings and booking statuses

- invoices and invoice items

- payment methods and facilities

### What We Need to Build

We are building a relational database \(SQLite\) that can store and manage the following information:

Hotels & Owners: Names, addresses, and star ratings\.

Rooms: Room types \(e\.g\., single, double\), pricing, amenities, and maximum occupancy\.

Customers: Addresses and personal contact details\.

Bookings: Timeframes, booking dates, and status \(e\.g\., "canceled"\)\.

Invoices: Payment methods \(Credit Card, Twint, etc\.\), currency, and line items \(like drinks or ski passes\)\.

### The Four Project Phases

Data Identification
In the first phase, we identified and structured the relevant business data required for the hotel reservation system\.

Data Modelling
In this phase, we created the Entity\-Relationship Diagram \(ERD\) using Visual Paradigm\.

Data Management
During this phase, the relational database was implemented in SQLite using SQL\.

Interaction
The final phase focused on user interaction through Python and Deepnote\.

### Implementation / User Stories

To prove our database works, we must solve specific "User Stories" using SQL queries:

As a User: We need to be able to search for hotels by city or stars, find available rooms, and book them\.

As an Admin: We need to be able to add/delete hotels and edit existing booking data\.
