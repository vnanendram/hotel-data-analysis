# Projectmanagement method

Kanban
The method is governed by a few simple and transformative rules:

- Visualize the Work: Our Work is represented by cards on a board, moving through the columns  "To Do," "In progress," and "Done\."

- Limit Work in Progress: By limiting how many tasks can be in the "Doing" column at once, we force the team to finish old tasks before starting new ones\. This prevents overload\.

- Manage Flow: Instead of watching the team, we watch the movement of the cards\. If cards are piling up in  a specific column you know exactly where the problem is\.
