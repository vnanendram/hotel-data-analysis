### 3\.3 Data Types and Constraints

This database uses optimised data types and constraints to ensure efficiency, data consistency and clear structure:

VARCHAR \(50–255\): For text values \(names, descriptions\)\. Lengths are tightly capped to save storage\.

INTEGER\(10\): For unique identifiers \(primary/foreign keys\) to link tables efficiently\.

BOOLEAN: For yes/no attributes \(e\.g\., pets\_allowed\)\.

DATE: For chronological data \(bookings, seasons\) to allow accurate filtering\.

DECIMAL\(10,2\): For precise monetary values \(e\.g\., prices in CHF\)\.

Data Quality:

Primary Key: Constraints uniquely identify each record\.

Foreign Keys: Constraints ensure valid relationships between tables\.

NOT NULL: Ensures mandatory fields cannot be left blank\.

UNIQUE: Prevents duplicate entries for sensitive fields like emails or passport numbers\.

Junction tables: Prevents duplicate Many\-to\-Many combinations through combined uniqueness rules\.

### 3\.4 Relationships

Relationships between entities are categorised as One\-to\-One, One\-to\-Many and Many\-to\-Many, and are described using \(min, max\) notation\.

There is a One\-to\-Many relationship between owner and hotel: owner \(1\) → hotel \(0\.\.\*\)                                                             

One owner can own multiple hotels, while each hotel belongs to exactly one owner\.

The hotel and address tables have a mandatory One\-to\-One: hotel \(1\.\.1\) ↔ address \(1\.\.1\)

Each hotel has exactly one address, and each address belongs to one hotel\.

In the customer domain, there is a mandatory One\-to\-One relationship between guest and address: guest \(1\.\.1\) ↔ address \(1\.\.1\)

Each guest has exactly one address, and each address belongs to one guest\.

There is also a One\-to\-Many relationship between hotel and room: hotel \(1\) → room \(0\.\.\*\)

One hotel can contain multiple rooms, while each room belongs to exactly one hotel\.

In the rooms section, there is a one\-to\-many relationship between room\_type and room: room\_type \(1\) → room \(0\.\.\*\)

One room type can be assigned to several rooms, while each room belongs to exactly one room type\.

In addition, the following applies between type\_code and room\_type: type\_code \(1\) → room\_type \(0\.\.\*\)

One type code can describe several room types, while each room type references exactly one type code\.

There is a One\-to\-Many relationship between guest and booking: guest \(1\) → booking \(0\.\.\*\)

One guest can create multiple bookings, while each booking belongs to exactly one guest\.

There is a One\-to\-Many relationship between room and booking: room \(1\) → booking \(0\.\.\*\)

A room can be booked more than once, but each booking refers to exactly one room\.

There is a One\-to\-Many relationship between booking\_status and booking: booking\_status \(1\) → booking \(0\.\.\*\)

One booking status can be used for multiple bookings, while each booking has exactly one status\.

There is an optional One\-to\-One relationship between a booking and an invoice: booking \(1\.\.1\) ↔ invoice \(0\.\.1\)

A booking may generate one invoice, but cancelled bookings may not have an invoice\.

In the invoicing module, there is a One\-to\-Many relationship between invoice and invoice\_item: invoice \(1\) → invoice\_item \(1\.\.\*\)

One invoice can contain multiple invoice items, while each invoice item belongs to exactly one invoice\.

There is also a One\-to\-Many relationship between payment\_method and invoice: payment\_method \(1\) → invoice \(0\.\.\*\)

One payment method can be used for multiple invoices, while each invoice uses exactly one payment method\.

There is a One\-to\-Many relationship between service\_type and invoice\_item: service\_type \(1\) → invoice\_item \(0\.\.\*\)

One service type can be used for multiple invoice items, while each invoice item references exactly one service type\.

There is a Many\-to\-Many relationship between hotel and hotel\_facility \(via a junction table\): hotel \(0\.\.\*\) ↔ hotel\_facility \(0\.\.\*\)

A hotel can offer multiple facilities, and the same facility can belong to multiple hotels\.

Finally, there is a Many\-to\-Many relationship between room and room\_facility \(via a join table\): room \(0\.\.\*\) ↔ room\_facility \(0\.\.\*\)

A room can contain multiple facilities, and the same facility can belong to multiple rooms\.

season\_code and room\_price also have a One\-to\-Many relationship: season\_code \(1\) → room\_price \(0\.\.\*\)

One season can contain several room prices, while each room price belongs to exactly one season\.

In addition, there is a One\-to\-Many relationship between room\_type and room\_price: room\_type \(1\) → room\_price \(0\.\.\*\)

One room type can have multiple seasonal prices, while each room price belongs to exactly one room type\.

### 3\.5 Normalisation 

Normalisation organises a database efficiently to avoid duplicate data and ensure accuracy\. In the specific hotel reservation system, the rules of the first, second, and third normal forms were applied for this purpose\. This ensures a clean, easily expandable structure and prevents errors when inserting, modifying or deleting data\. 

1\.Normal Form \(1NF\): Atomicity
A database table achieves First Normal Form \(1NF\) by meeting three core criteria: all attributes must contain atomic, single values, there must be no repeating groups and each record must be uniquely identifiable through a primary key\.

In this project, all tables fully comply with 1NF requirements\. Every entity is assigned a distinct primary key, such as hotel\_id\_PK, guest\_id\_PK, room\_id\_PK and booking\_id\_PK to ensure unique record identification\. 

Furthermore, multi\-value data is strictly avoided within single attributes\. For instance, rather than storing room facilities as a list inside the main room table, they are isolated in a dedicated room\_facility table and linked using the room\_has\_facility junction table\. 

A parallel approach is utilised for hotels, where hotels and their respective features are connected through a hotel\_has\_facility table\. By eliminating repeating groups and breaking down data into atomic components, this design guarantees a highly flexible and scalable database structure\.

2\. Second Normal Form \(2NF\): Full functional dependency
A database table achieves Second Normal Form \(2NF\) if it already fulfills all the requirements of 1NF and ensures that all non\-key attributes depend entirely on the primary key\. This rule specifically eliminates partial dependencies, meaning no non\-key attribute can depend on only a part of a composite primary key\.

In this project, the database structure successfully fulfills 2NF because all attributes are fully dependent on their corresponding primary keys\. For example, room information depends entirely on room\_id\_PK, booking information relies solely on booking\_id\_PK and invoice details are fully dependent on invoice\_id\_PK\. 

To maintain this standard, the design separates related information into dedicated entities\. Room types are stored independently in the room\_type table, room categories are standardised using a type\_code and payment methods are managed in a separate payment\_method table\. This methodical isolation avoids partial dependencies and significantly improves overall data organization\.

3\. Third Normal Form \(3NF\): No transitive dependencies
A table achieves Third Normal Form \(3NF\) when it meets all 2NF criteria and ensures no non\-key attribute depends on another non\-key attribute and eliminating transitive dependencies\. The hotel reservation system fulfills 3NF by isolating reusable business information into independent tables\.

To prevent data duplication and redundancy, address information is moved to a dedicated address table referenced via foreign keys and room prices are separated into a room\_price table linked to types and seasons\. Data consistency is enforced by isolating booking statuses in a booking\_status table to eliminate varied text formatting and by storing invoice categories in a service\_type table to avoid repeating descriptions like "Accommodation" or "Food"\. Furthermore, payment methods are managed in a distinct payment\_method table for easier future extensions\. 

Finally, Many\-to\-Many relationships are resolved using the hotel\_has\_facility and room\_has\_facility junction tables, preventing multi\-value columns and ensuring a fully normalised structure\.


Protection Against Database Anomalies

Insertion Anomaly: The normalised database structure effectively prevents common database anomalies and specifically insertion anomalies\. This design ensures that new facilities payment methods or booking statuses can be added to the system independently without requiring the prior creation of hotels, bookings or invoices\. 

For example, a new payment method can be inserted into the system on its own and a new room facility can exist in the database before ever being assigned to a specific room\.

Update Anomaly: The normalised database structure also eliminates update anomalies by ensuring that data only needs to be modified in a single location\. 

For example, if the name of a payment method changes, it only requires an update within the payment\_method  table\. Because all related invoices reference this single record, they automatically reflect the updated value\. This centralised approach guarantees data consistency across the entire database and prevents conflicting information\.

Deletion anomaly \(Delete\): The normalised database structure successfully prevents delete anomalies by ensuring that removing transactional data does not automatically wipe out crucial master data\. 

For example, deleting a booking record does not delete the associated guest from the system and deleting an invoice does not remove the underlying payment methods or service types\. 

## 4\. Implementation 

### 4\.1 Database Structure \(SQLite\) 

The database was implemented in SQLite based on an Entity\-Relationship \(ER\) model created in Visual Paradigm\. Following a relational database structure, the system uses tables, primary keys, and foreign keys to connect entities, organising them into three distinct functional groups\.

The master data tables store foundational, reusable information such as addresses, guests, hotels, room types, booking statuses, payment methods and service types\. 

Transaction tables capture core business processes and activities, including bookings, invoices and individual invoice items\. 

Finally, junction tables, such as hotel\_has\_facility and room\_has\_facility are utilised to cleanly resolve Many\-to\-Many relationships\.

Throughout this architecture, primary keys uniquely identify each record, while foreign keys enforce referential integrity to ensure relationships between tables remain consistent\.

This organised structure allows the database to store hotel, room, booking and invoice data in a clear, reliable, and standardised manner\.

### 4\.2 Test Data 

Realistic test data was inserted into the database to verify the correct functionality of the system and to ensure that all user stories can be successfully executed\. This dataset simulates actual hotel reservation processes by including authentic examples of Swiss hotel locations like Davos, St\. Moritz, Gstaad and Interlaken, alongside various hotel categories and star ratings\. 

It also features several room types, such as single rooms, double rooms, studios, family rooms, and suites, along with their respective room and hotel facilities\. Furthermore, the data encompasses guests with full contact details, bookings reflecting different operational statuses and invoices complete with specific payment methods and line items\.

The primary purpose of this test data is to evaluate whether essential system operations, including searches, bookings, availability checks and invoice queries, return accurate results\. By mimicking real world scenarios, the dataset also serves as a crucial diagnostic tool to identify and resolve any potential errors in table relationships, foreign key constraints or SQL queries prior to final submission\.

### 4\.3 Joins

JOIN operations are essential to this project because most user stories require retrieving information distributed across several related tables\. 

For example, generating hotel search results requires combining data from the hotel, address, room, room\_type and room\_price tables\. 

Similarly, pulling detailed room information relies on joins between room, room\_type, type\_code, room\_has\_facility and room\_facility\. 

Accessing complete invoice information likewise requires joining the invoice, invoice\_item, payment\_method, service\_type and booking tables\.

The extensive use of JOIN operations demonstrates the core strength of a relational database structure\. Instead of storing all information in a single, bloated table, data is divided into distinct, logical entities and dynamically connected through SQL queries only when needed\. This approach significantly improves data quality, eliminates redundancy and ensures the database structure remains flexible and scalable\.

### 4\.4 SQL Queries \(User Stories\)  

The SQL queries implemented in this project demonstrate how the hotel reservation system fulfills the user stories defined in the project specification, showcasing how relational databases manage, retrieve and manipulate business data within a realistic environment\. The implementation spans both customer\-oriented and administrator\-oriented functionalities, utilising various SQL concepts such as SELECT, INSERT, UPDATE, DELETE, JOIN, GROUP BY, ORDER BY and subqueries\.

As a User

From the user perspective, the system supports typical reservation activities\. Customers can search for hotels by city, filter by star ratings, search for rooms based on guest count and check real\-time availability for specific dates\. 

They can also view detailed room profiles including types, facilities and prices and book available rooms\. To support these actions, queries dynamically connect multiple tables via JOIN operations, such as combining hotel, address, room, and pricing data, while filtering out existing reservations during availability checks\.

As an Admin User

For administrative operations, the system enables the management of hotels, bookings, customer information, room data and invoices\. Administrator related queries allow users to insert new hotels, update hotel information and remove hotels from the database\. 

Admins can also display all bookings across all hotels, update missing customer details like phone numbers and manage booking and invoice records\. These operations primarily use INSERT, UPDATE and DELETE statements combined with relational constraints to maintain strict data consistency and integrity\.

Additional Useful Queries

Beyond the core user stories, several additional queries were implemented to highlight the analytical capabilities of the relational system\. These include displaying unpaid invoices, calculating hotel revenue, listing hotel facilities, analysing booking trends and displaying invoice details\. Together, these queries validate that the database structure successfully supports all required business processes while ensuring scalability and efficient data retrieval\.

### 4\.5 Python User Interaction \(Deepnote\) 

Python was utilised within Deepnote to build a simple interaction layer between the user and the SQLite database, making data management significantly more user\-friendly\. Instead of writing manual SQL queries, users can simply input specific values, such as a city, number of guests or travel dates and Python automatically executes the corresponding SQL query to display the results in a structured format\.

The Python implementation supports core user actions, including searching hotels by city, filtering by star ratings, checking room availability, displaying room details, creating bookings and viewing booking or invoice data\. 

This integration relies on two key Python libraries: sqlite to establish the connection with the SQLite database file, pandas to process and display the query results cleanly in dataframes or tables\. 

Ultimately, this interaction layer demonstrates how relational databases can be seamlessly connected to practical workflows, presenting clear and actionable data for business users\.

## 7\. Reflection & Conclusion 

### 7\.1 What went well? 

Overall, the project developed very positively, allowing the team to successfully apply theoretical concepts learned during the module within a practical context\. One of the greatest strengths of this project was its structured development process\. 

The creation of the ER model went particularly well following the initial planning phase, as the overall database structure became increasingly clear once the main entities and relationships were identified\. 

The SQL implementation was another highly successful aspect, with the team gaining significant confidence in writing queries and working with relational databases throughout the semester\.

Group collaboration also contributed immensely to the success of the project\. Additionally, utilising Deepnote proved highly advantageous\. Because Deepnote combines SQL, Python, and documentation into a single workspace, the team was able to collaborate efficiently without the friction of switching between multiple tools, which greatly streamlined continuous development and testing\.

### 7\.2 What was challenging? 

Despite the positive overall outcome, the project also presented several challenges\. One of the most difficult parts was the initial design of the ER model\. It was sometimes difficult to decide how entities should be connected, which attributes belonged to which table and how to correctly model relationships\.

The implementation of complex SQL queries also became challenging at certain points\. Queries involving several JOIN operations required a precise understanding of the relationships between tables\. In particular, the availability checks for rooms were more difficult than expected because overlapping booking dates had to be considered correctly\.

Another challenge was maintaining consistent and realistic test data\. Because the database contains many foreign key relationships, the order in which records were inserted into the tables was extremely important\. Even small mistakes such as incorrect IDs or missing references could lead to errors that affected multiple tables\. This required careful attention and systematic testing\.

The topic of normalisation also required additional effort\. Although the theoretical principles seemed understandable at first, applying them correctly in practice was more complex\. 

### 7\.3 What did we learn? 

The project provided valuable insights into both relational databases and collaborative software development\. One of the most important learning outcomes was gaining a deeper understanding of how relational database systems are designed and implemented in practice\. Through hands\-on application, core concepts such as entities, relationships, cardinalities, primary keys, foreign keys, and normalisation became much clearer than through theory alone\.

Additionally, the project significantly improved the team's SQL skills\. While some queries initially seemed difficult and complex, regular practice taught the team how to structure queries logically and troubleshoot problems systematically\. The process also highlighted the critical importance of proper database planning, demonstrating that a well\-designed ER model simplifies many later steps, including query writing, testing and system maintenance\.

### 7\.4 Conclusion

The Hotel Reservation System project successfully merged theoretical knowledge with implementation, offering a deep understanding of the relational database development lifecycle\. While early phases like ER modeling and complex SQL queries proved challenging, working through these design problems ultimately enhanced the team's technical confidence and problem\-solving abilities\. 

The final outcome is a fully functional relational database system that satisfies all user stories and demonstrates the core principles of business applications, effectively strengthening both the technical and collaborative skills of the team\.
