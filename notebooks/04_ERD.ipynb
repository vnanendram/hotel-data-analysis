# ERD

The following section contains the technical overview of the entities used in the hotel reservation system\. The Entity\-Relationship Diagram \(ERD\) visualises the database structure and the relationships between all entities\.

The ER model was created in Visual Paradigm and later implemented in SQLite\.

The database consists of:

Master data tables

Transaction tables

Junction tables for Many\-to\-Many relationships

The final database structure consists of a comprehensive set of entities that together represent the core processes of a hotel reservation system\. 

The entity structure was carefully designed to support the requirements defined in the user stories\. Master data entities provide reusable reference information, transaction entities store operational activities and junction entities enable flexible Many\-to\-Many relationships without introducing redundancy\.

The use of primary keys ensures that every record can be uniquely identified, while foreign keys maintain the relationships between entities and guarantee referential integrity\. Additional constraints such as NOT NULL, UNIQUE and foreign key constraints contribute to data quality and consistency throughout the database\.

The final ER model contains entities for addresses, owners, hotels, guests, bookings, rooms, facilities, room pricing, invoices and payment methods\. Together, these entities create a complete and realistic representation of a hotel reservation environment\.

Overall, the database design provides a scalable, maintainable and fully normalised solution that successfully supports the business requirements of the hotel reservation system while demonstrating the practical application of relational database concepts learned throughout the module\.
